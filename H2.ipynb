Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là độ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn hình

```python
a=int(input())
b=int(input())  
c=int(input())
if a+b>c and a+c>b and b+c>a:
    if a==b==c:
        print("Tam giác đều")
    elif a==b or a==c or b==c:
        print("Tam giác cân")
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        print("Tam giác vuông")
    else:
        print("Tam giác thường")
else:
    print("Khong la ba canh cua mot tam giac")

```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
import math

x0,y0=map(int, input("Nhập tọa độ tâm I: ").split())
a,b=map(int, input("Nhập tọa độ điểm A: ").split())
R=int(input("Nhập bán kính R: "))
d=math.sqrt((x0 - a)**2 + (y0 - b)**2)
if d < R:
    print("Điểm A nằm trong hình tròn")
else:
    print("Điểm A không nằm trong hình tròn")
```

```python
import math

x0,y0 = map(float, input().split())
r = float(input())

x,y = map(float, input().split())

d = math.sqrt((x - x0)**2 + (y - y0)**2)

print(f"\nKhoảng cách từ A đến tâm I là: {d:.2f}")

if math.isclose(d, r):
    print("Kết quả: Điểm A thuộc đường tròn.")
elif d < r:
    print("Kết quả: Điểm A nằm trong đường tròn.")
else:
    print("Kết quả: Điểm A nằm ngoài đường tròn.")
```
